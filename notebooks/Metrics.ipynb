```python
import numpy as np
from matplotlib import pyplot as plt

filter_times = [result[2] for result in filter_results]
rf_times = [result[2] for result in rf_results]
dl_times = [result[2] for result in dl_results]

mean_times = [np.mean(filter_times), np.mean(rf_times), np.mean(dl_times)]
std_times = [np.std(filter_times), np.std(rf_times), np.std(dl_times)]

method_names = ['Filter', 'Random Forest', 'Deep Learning (GPU)']

plt.figure(figsize=(12, 7))
bars = plt.bar(method_names, mean_times, yerr=std_times, capsize=5, color=['skyblue', 'lightgreen', 'salmon'], edgecolor='black')

plt.title('Time Analysis of Segmentation Methods', fontsize=16)
plt.ylabel('Execution Time (seconds)', fontsize=12)
plt.xlabel('Method', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, f'{yval:.2f} ± {std_times[bars.index(bar)]:.2f}s', va='bottom', ha='center', fontsize=10)

plt.show()
```

```python
import pandas as pd
from util.Evaluate import Evaluate

pd.set_option('display.float_format', '{:.4f}'.format)

all_metrics_data = []

mask_paths = [item[3] for item in filtered_dataset]
mask_images = ImageLoader.load_images(mask_paths)

for (base_name, _, _, _), filter_res, rf_res, dl_res, label, mask in zip(filtered_dataset, filter_results, rf_results, dl_results, manuals, mask_images):

    target_shape = (512, 512) 

    y_true_resized = ImagePreprocessing.resize(label, target_shape)
    y_true = (y_true_resized > 0).astype(np.uint8)

    mask_resized = ImagePreprocessing.resize(mask, target_shape)
    if len(mask_resized.shape) == 3:
        mask_resized = mask_resized[:,:,0]  

    _, filter_map, _ = filter_res
    y_pred_filter = (filter_map > 0).astype(np.uint8)

    _, rf_map, _ = rf_res
    y_pred_rf = rf_map.astype(np.uint8)

    _, dl_map, _ = dl_res
    dl_map_resized = ImagePreprocessing.resize(dl_map, target_shape)
    y_pred_dl = (dl_map_resized > 0.5).astype(np.uint8)

    metrics_for_image = []

    accuracy_filter = Evaluate.accuracy(y_true, y_pred_filter, mask_resized)
    sensitivity_filter = Evaluate.sensitivity(y_true, y_pred_filter, mask_resized)
    specificity_filter = Evaluate.specificity(y_true, y_pred_filter, mask_resized)
    cm_filter = Evaluate.confusion(y_true, y_pred_filter, mask_resized)
    
    tp, tn, fp, fn = cm_filter['TP'], cm_filter['TN'], cm_filter['FP'], cm_filter['FN']
    precision_filter = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall_filter = sensitivity_filter  
    f1_filter = 2 * (precision_filter * recall_filter) / (precision_filter + recall_filter) if (precision_filter + recall_filter) > 0 else 0.0
    
    metrics_filter = {
        'accuracy': accuracy_filter,
        'precision': precision_filter,
        'f1': f1_filter,
        'specificity': specificity_filter,
        'sensitivity': sensitivity_filter,
        'Method': 'Filter'
    }
    metrics_for_image.append(metrics_filter)

    accuracy_rf = Evaluate.accuracy(y_true, y_pred_rf, mask_resized)
    sensitivity_rf = Evaluate.sensitivity(y_true, y_pred_rf, mask_resized)
    specificity_rf = Evaluate.specificity(y_true, y_pred_rf, mask_resized)
    cm_rf = Evaluate.confusion(y_true, y_pred_rf, mask_resized)
    
    tp, tn, fp, fn = cm_rf['TP'], cm_rf['TN'], cm_rf['FP'], cm_rf['FN']
    precision_rf = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall_rf = sensitivity_rf
    f1_rf = 2 * (precision_rf * recall_rf) / (precision_rf + recall_rf) if (precision_rf + recall_rf) > 0 else 0.0
    
    metrics_rf = {
        'accuracy': accuracy_rf,
        'precision': precision_rf,
        'f1': f1_rf,
        'specificity': specificity_rf,
        'sensitivity': sensitivity_rf,
        'Method': 'Random Forest'
    }
    metrics_for_image.append(metrics_rf)

    accuracy_dl = Evaluate.accuracy(y_true, y_pred_dl, mask_resized)
    sensitivity_dl = Evaluate.sensitivity(y_true, y_pred_dl, mask_resized)
    specificity_dl = Evaluate.specificity(y_true, y_pred_dl, mask_resized)
    cm_dl = Evaluate.confusion(y_true, y_pred_dl, mask_resized)
    
    tp, tn, fp, fn = cm_dl['TP'], cm_dl['TN'], cm_dl['FP'], cm_dl['FN']
    precision_dl = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall_dl = sensitivity_dl
    f1_dl = 2 * (precision_dl * recall_dl) / (precision_dl + recall_dl) if (precision_dl + recall_dl) > 0 else 0.0
    
    metrics_dl = {
        'accuracy': accuracy_dl,
        'precision': precision_dl,
        'f1': f1_dl,
        'specificity': specificity_dl,
        'sensitivity': sensitivity_dl,
        'Method': 'Deep Learning'
    }
    metrics_for_image.append(metrics_dl)
    
    all_metrics_data.extend(metrics_for_image)

    print(f"\n📊 Metrics Table for: {base_name}")
    df_image = pd.DataFrame(metrics_for_image).set_index('Method')
    display(df_image)

    print(f"\nConfusion Matrices for {base_name}:")
    print("Filter:")
    Evaluate.print_confusion_matrix(y_true, y_pred_filter, mask_resized)
    print("\nRandom Forest:")
    Evaluate.print_confusion_matrix(y_true, y_pred_rf, mask_resized)
    print("\nDeep Learning:")
    Evaluate.print_confusion_matrix(y_true, y_pred_dl, mask_resized)
    print("-" * 60)

print("\n\n" + "="*60)
print("📊 Overall Average Metrics Across All Test Images (Masked Evaluation)")
print("="*60)

df_all = pd.DataFrame(all_metrics_data)
df_summary = df_all.groupby('Method').mean()

display(df_summary)
```
